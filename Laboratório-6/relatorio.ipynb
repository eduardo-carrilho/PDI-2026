## 1. Introdução
A limiarização (*thresholding*) é uma das técnicas mais fundamentais em Processamento Digital de Imagens para a segmentação visual. O objetivo deste procedimento é separar os objetos de interesse do fundo da imagem, convertendo uma imagem em tons de cinzento para uma imagem binária (onde os píxeis assumem apenas os valores 0 ou 255). 

Neste laboratório, exploramos e comparamos métodos de limiarização manual (baseada na inspeção visual de histogramas) e métodos automáticos (como o Algoritmo de Otsu), bem como a limiarização adaptativa. Aplicamos estes conceitos em imagens estáticas (fotos e avatares da equipa), em processamento de vídeo em tempo real através da webcam, e finalmente, nas imagens do nosso Projeto Final (Corretor de Gabaritos/Testes OMR).

## 3. Análise e Discussão dos Estudos Realizados

### Análise Prática dos Experimentos (Atividades 3.a e 3.b)
Ao aplicarmos a limiarização nos nossos avatares e no vídeo da webcam, notámos que o limiar manual requer ajustes constantes sempre que a iluminação da sala é alterada. Durante a gravação com a webcam, observámos certos artefactos luminosos: as sombras do ambiente (como a sombra dos nossos próprios corpos) eram frequentemente lidas como "objetos" escuros num limiar manual, unindo-se aos nossos contornos. O método automático de Otsu comportou-se melhor, recalculando o contraste ideal dinamicamente.

### Respostas às Questões de Análise

**(5.a) Diferença entre Inspeção Visual do Histograma e Métodos Automáticos:**
* **Inspeção Visual:** Consiste em analisar o gráfico do histograma para tentar encontrar manualmente o "vale" entre dois picos que separam o fundo do objeto. 
  * *Vantagens:* É rápida para imagens perfeitamente bimodais com alto contraste.
  * *Desvantagens:* É subjetiva, requer intervenção humana e falha redondamente quando a iluminação gera vales mal definidos. Inviável para automação em vídeos contínuos.
* **Métodos Automáticos (Otsu):** Utilizam modelação estatística para calcular o ponto exato que maximiza a variância interclasses.
  * *Vantagens:* Totalmente autónomo e objetivo. Estatisticamente é o melhor método para separar duas classes dominantes de píxeis.
  * *Desvantagens:* Se a imagem for unimodal ou tiver um fundo complexo (onde o objeto representa apenas uma fração minúscula do histograma), o cálculo da variância não consegue encontrar uma separação válida.

**(5.b) Impacto da Escolha de Limiares (img2):**
* **Limiar muito baixo (T=50):** Valores muito baixos forçam o algoritmo a classificar como "objeto" apenas os píxeis extremamente escuros. Isto pode fragmentar o objeto ou criar buracos onde o objeto não é 100% negro, resultando numa segmentação incompleta.
* **Limiar muito alto (T=125):** Valores muito altos engolem a fronteira entre o fundo e o objeto, fundindo sombras ligeiras com o objeto real. Isto "engorda" a silhueta ou gera lixo visual (falsos positivos).
* **Sugestão de Limiar Ideal:** Com base na transição suave do histograma entre as zonas escuras e claras, a inspeção sugere que o valor de mínimo local (o vale entre os dois picos) aproxima-se de **T=68**. Este valor corta exatamente a transição, preservando a forma real.

**(5.c) Vantagem do método de Otsu (Imagem "rice.png"):**
A imagem de grãos de arroz possui um sério problema de gradiente de luz (iluminação não uniforme). A técnica de tentativa e erro frustra o utilizador, porque ao tentar ajustar o limiar para detetar os grãos no fundo escuro inferior, perdem-se os grãos na parte superior iluminada. A principal vantagem do método automático de Otsu é a sua capacidade de eliminar o viés visual humano, aplicando um limiar matemático ideal calculado sobre a distribuição global que preserva o máximo de grãos possíveis num compromisso ótimo de variância.

## 4. Conclusões
O Laboratório 6 permitiu consolidar a diferença crítica entre o processamento baseado na perceção visual humana e a otimização estatística das máquinas. 

Constatamos que os métodos globais (como o de Otsu) são excelentes em ambientes controlados, porém sofrem variações extremas em vídeos de webcam ou fotos mal iluminadas. A execução destes exercícios elucidou a nossa equipa sobre a importância do pré-processamento no nosso Trabalho Final. No sistema do Corretor de Gabaritos, compreendemos que para ler as bolhas com eficácia, a combinação de filtros de equalização de luz (CLAHE) com binarizações precisas aliadas a operadores morfológicos é imperativa para evitar o "efeito donut" ou rasuras causadas pela iluminação do ambiente.
