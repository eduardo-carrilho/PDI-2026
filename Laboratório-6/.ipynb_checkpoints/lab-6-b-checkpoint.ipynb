```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def LimiarManual(img):
    img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    c = (img > 150).astype("uint8") * 255
    return cv2.cvtColor(c, cv2.COLOR_GRAY2BGR)
    
def LimiarAutomatico(img):
    img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    T,c = cv2.threshold(img,0,255,cv2.THRESH_OTSU) 
    return cv2.cvtColor(c, cv2.COLOR_GRAY2BGR)
    
cap = cv2.VideoCapture(0)

# Get current width of frame
width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)   # float
# Get current height of frame
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT) # float
# Define Video Frame Rate in fps
fps = 30.0

# Define the codec and create VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter('saida.avi', fourcc, fps, (int(width),int(height)) )

threshold_type = 'auto'

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break

    if threshold_type == 'auto':
        frame_limiar = LimiarAutomatico(frame)
    else:
        frame_limiar = LimiarManual(frame)
    frames_concat = [frame,frame_limiar]
    concat = cv2.hconcat(frames_concat)
    cv2.imshow('frame', concat)
    key = cv2.waitKey(1) 
    if key == ord('c'):
        if threshold_type == 'auto':
            threshold_type = 'manual'
        else:
            threshold_type = 'auto'
    elif key == ord('q'):
        break

# Release everything if job is finished
cap.release()
out.release()
cv2.destroyAllWindows()
```
