```python
img = cv2.imread('foto-geral.jpg',0)
img_avatar = cv2.imread('foto_avatar.png',0)
```

## Limiar 50

```python
limiar = 50

plt.figure(figsize=(15,10))
plt.subplot(231),plt.imshow(img,"gray"),plt.title('Entrada Foto Geral')
plt.subplot(232),plt.hist(img.ravel(), bins=256, range=[0,256]),plt.title('Histograma Foto Geral')
plt.subplot(233),plt.imshow(img>limiar,"gray"),plt.title(f'Limiar = {limiar}')
plt.subplot(234),plt.imshow(img_avatar,"gray"),plt.title('Entrada Foto Avatares')
plt.subplot(235),plt.hist(img_avatar.ravel(), bins=256, range=[0,256]),plt.title('Histograma Foto Avatares')
plt.subplot(236),plt.imshow(img_avatar>limiar,"gray"),plt.title(f'Limiar = {limiar}')
plt.show()
```

```python
# Usando limiarização de OTSU: cv2.THRESH_OTSU
#
#f = cv2.imread('foto-geral.jpg',0)

#f = f[0:585,0:650]
T,c = cv2.threshold(img,0,255,cv2.THRESH_OTSU) #<<<<<<<<<<
print(T)
T2, c2 = cv2.threshold(img_avatar,0,255,cv2.THRESH_OTSU)
plt.figure(figsize=(8,8))
plt.subplot(221),plt.imshow(img,"gray"),plt.title('Input Foto Geral')
plt.subplot(222),plt.imshow(c,"gray"),plt.title('Thresholding Otsu T='+str(T)+' Foto Geral')
plt.subplot(223),plt.imshow(img_avatar,"gray"),plt.title('Input Foto Avatares')
plt.subplot(224),plt.imshow(c2,"gray"),plt.title('Thresholding Otsu T='+str(T2)+' Foto Avatares')
plt.show()
```
