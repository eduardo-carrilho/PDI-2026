# 2. Detecção de Bordas

A detecção de bordas é um processo fundamental em processamento de imagens, pois permite identificar **transições bruscas de intensidade**, geralmente associadas a contornos de objetos.

Nesta etapa foram utilizados três métodos distintos.

---

# 2.1 Filtro Laplaciano

O **Laplaciano** é um operador diferencial baseado na **segunda derivada da imagem**.

Ele destaca regiões onde ocorre mudança rápida de intensidade, sendo eficaz para identificar bordas em todas as direções.

Matematicamente, o operador Laplaciano pode ser representado como:

\[
\nabla^2 f = \frac{\partial^2 f}{\partial x^2} + \frac{\partial^2 f}{\partial y^2}
\]

No código, o filtro foi aplicado utilizando a função:

```python
cv.Laplacian()
```

Esse operador gera uma nova imagem onde as bordas aparecem com maior intensidade.

---

# 2.2 Filtros Sobel

Os filtros **Sobel** são detectores de bordas baseados no cálculo do **gradiente da imagem**.

Foram aplicadas duas variações:

- **Sobel X** – detecta bordas verticais  
- **Sobel Y** – detecta bordas horizontais  

As operações foram realizadas com a função:

```python
cv.Sobel()
```

Além disso, foram testados **diferentes tamanhos de kernel**, como:

- `ksize = 3`
- `ksize = 9`

A variação do kernel altera o nível de suavização e a sensibilidade do detector de bordas.

Kernels maiores tendem a produzir bordas mais suaves e menos sensíveis a ruídos.

# 3. Detector de Bordas Canny

O **Canny Edge Detector** é um dos algoritmos mais utilizados para detecção de bordas, pois produz resultados mais limpos e robustos.

O algoritmo consiste em várias etapas:

1. Suavização da imagem
2. Cálculo do gradiente
3. Supressão de não-máximos
4. Aplicação de limiares duplos
5. Rastreamento de bordas

No código, o detector foi aplicado utilizando:

```python
cv.Canny()
```

Foram testadas diferentes combinações de limiares:

- `(50, 150)`
- `(100, 200)`
- `(200, 250)`

Esses valores influenciam diretamente a quantidade de bordas detectadas:

- **Limiar baixo** → detecta mais bordas (inclusive ruídos)
- **Limiar alto** → detecta apenas bordas mais fortes

A análise visual permitiu comparar o impacto desses parâmetros na detecção de bordas.

# 4. Realce de Nitidez da Imagem

Além da detecção de bordas, também foram aplicadas técnicas de **realce de nitidez**, com o objetivo de enfatizar detalhes da imagem.

Foi utilizada uma imagem com **ruído gaussiano**, permitindo observar melhor os efeitos dos métodos.

---

# 4.1 Realce com Laplaciano

O realce de nitidez utilizando Laplaciano baseia-se na ideia de **adicionar ou subtrair a informação de alta frequência da imagem original**.

A fórmula utilizada pode ser representada como:

\[
g(x,y) = f(x,y) - c \cdot \nabla^2 f(x,y)
\]

Onde:

- \(f(x,y)\) é a imagem original  
- \(\nabla^2 f(x,y)\) é o Laplaciano  
- \(c\) é um fator de controle  

No código, o Laplaciano foi aplicado para gerar uma versão **mais nítida da imagem original**.

O resultado evidencia melhor os contornos e detalhes da imagem.

---

# 4.2 Filtro Highboost

O **Highboost Filtering** é uma técnica de realce baseada na amplificação das componentes de alta frequência.

A fórmula geral é:

\[
g(x,y) = A \cdot f(x,y) - f_{suavizada}(x,y)
\]

Onde:

- \(A\) é o fator de amplificação
- \(f_{suavizada}\) é a imagem após aplicação de um filtro passa-baixa

No experimento foram utilizados diferentes níveis de amplificação:

- \(k = 1.5\)
- \(k = 3.0\)

Valores maiores de \(k\) aumentam o realce dos detalhes, porém podem também amplificar ruídos.
