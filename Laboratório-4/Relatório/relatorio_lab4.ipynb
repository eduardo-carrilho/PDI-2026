## 1. Introdução
Este relatório apresenta as atividades do Laboratório 4, voltadas ao estudo de **filtragem de imagens** utilizando a biblioteca **_OpenCV_** em Python. O objetivo é compreender o funcionamento de **filtros passa-baixa**, usados para suavização e redução de ruído, e **passa-alta**, utilizados para detecção de bordas e aumento de nitidez. Também são analisados os efeitos de diferentes tipos de ruído e tamanhos de kernel nas imagens, além da aplicação das técnicas em tempo real com webcam.
