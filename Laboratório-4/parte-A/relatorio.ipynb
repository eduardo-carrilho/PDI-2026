## 1. Introdução
Este relatório detalha as atividades do Laboratório 4, focadas na aplicação de **filtros passa-baixa** para suavização de imagens e redução de ruído utilizando a biblioteca **OpenCV**. O objetivo é compreender como diferentes algoritmos de convolução espacial (Média, Gaussiano, Mediana e Bilateral) impactam a qualidade visual e a preservação de detalhes, além de avaliar quantitativamente a recuperação de imagens ruidosas através da métrica PSNR.

---

## 2. Procedimentos Experimentais

Para a realização das atividades, os procedimentos foram divididos em etapas principais, conforme armazenado nos ficheiros `.ipynb`:

### 2.1. Filtragem de Imagens Originais (1-a.ipynb)
Implementação de rotinas para aplicar os quatro tipos de filtros (`blur`, `GaussianBlur`, `medianBlur` e `bilateralFilter`) em imagens limpas (`foto-geral.jpg` e `foto_avatar.png`). Foram utilizados tamanhos de kernel progressivos: **3x3, 5x5, 11x11 e 29x29**, com os resultados salvos no diretório `../imagens/`.

### 2.2. Simulação de Ruído Severo (1-b.ipynb)
Criação de imagens de teste através da inserção artificial de ruídos para validar a eficácia dos filtros:
* **Ruído Gaussiano:** Adicionado à imagem original com média 0 e desvio padrão (std) de 30.
* **Ruído Sal e Pimenta:** Inserido com uma proporção de ruído de 0.2 (20%), resultando numa degradação severa da matriz de pixels.

### 2.3. Avaliação de Métricas e Restauração (1-c.ipynb)
Cálculo do **PSNR (Peak Signal-to-Noise Ratio)** para comparar a imagem original com a versão recuperada após a filtragem. Esta etapa permitiu identificar objetivamente qual filtro oferece a melhor restauração para cada cenário de ruído.

### 2.4. Processamento em Tempo Real (2-c.ipynb)
Desenvolvimento de um sistema de filtragem contínua via webcam (dispositivo 0). O código aplica simultaneamente o "melhor" filtro (Mediana 5x5) e o "pior" filtro (Média 29x29) identificados, permitindo a captura de frames processados através da tecla 's'.

## 3. Análise e Discussão dos Resultados

### 3.1. Impacto do Tamanho do Kernel
Observou-se que o aumento do tamanho do kernel é diretamente proporcional ao nível de suavização (borramento) da imagem. Enquanto kernels de 3x3 e 5x5 removem ruídos leves mantendo bordas discerníveis, o kernel de 29x29 causa uma perda massiva de informação, transformando a imagem num conjunto de manchas de cores.

### 3.2. Comparação de Filtros contra Ruído
Conforme os testes realizados, os resultados demonstraram especializações claras para cada algoritmo:

* **Filtro de Mediana:** Mostrou-se superior para o ruído **Sal e Pimenta**. Ao selecionar o valor central da vizinhança, ele elimina eficazmente os pixels espúrios sem "espalhar" o erro pela imagem. O PSNR obtido para este filtro foi de **33.19 dB**.
* **Filtro Bilateral:** Demonstrou ser eficaz na suavização de cores mantendo as linhas de contorno nítidas, devido à sua característica de considerar tanto a distância espacial quanto a diferença de intensidade entre pixels.
* **Filtro de Média (Blur):** Apresentou o desempenho mais simples, reduzindo o ruído mas degradando severamente as bordas e o contraste da imagem original.

## 4. Conclusão

A realização deste laboratório permitiu consolidar a compreensão sobre o domínio espacial das imagens e a manipulação de frequências. Conclui-se que:
* **Adequação do Filtro:** A escolha da técnica deve ser guiada pelo tipo de ruído. O filtro de mediana é indispensável para ruídos impulsivos, enquanto filtros Gaussianos são adequados para ruídos de sensor.
* **Trade-off de Nitidez:** Existe um limite crítico para a suavização. O uso de kernels excessivos (29x29) destrói a utilidade da imagem para reconhecimento de padrões.
* **Métricas Objetivas:** O uso do PSNR é fundamental para validar tecnicamente se o processamento está a recuperar a informação original ou apenas a mascarar a degradação.
