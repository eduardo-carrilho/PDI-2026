## 1. Introdução
No processamento digital de imagens, a morfologia matemática atua como uma ferramenta poderosa para a extração de componentes visuais fundamentais, permitindo a representação e descrição da forma de regiões, como fronteiras e esqueletos. Estas abordagens são frequentemente aplicadas em etapas de pré e pós-processamento, englobando operações de filtragem morfológica, preenchimento e limpeza visual.

Neste laboratório, as bases da morfologia (especificamente as operações de **Erosão**, **Dilatação**, **Abertura** e **Fechamento**) foram testadas na prática. Por meio das funções nativas do OpenCV, avaliou-se o impacto destas operações quando combinadas a diferentes Elementos Estruturantes (Kernels), além de explorar o comportamento do Gradiente Morfológico na extração de fronteiras e contornos.

## 2. Procedimentos Experimentais
Os experimentos fundamentaram-se na aplicação sistemática das transformações morfológicas usando Python e a biblioteca OpenCV. O fluxo principal de execução seguiu as seguintes etapas:

* **Carregamento e Binarização:** As imagens foram convertidas para escala de cinza e posteriormente binarizadas utilizando limiarização (como o algoritmo de Otsu via `cv2.threshold`). Esta conversão foi essencial, visto que as operações morfológicas requerem imagens binárias para atuarem corretamente sobre as formas.
* **Definição dos Kernels:** Com a função auxiliar `cv2.getStructuringElement()`, foram definidos Elementos Estruturantes de diferentes topologias (retangulares, em formato de cruz e elípticos) e vários tamanhos, variando de 3x3 até 21x21.
* **Erosão e Dilatação:** Através de `cv2.erode()` e `cv2.dilate()`, estas operações foram aplicadas em imagens com o propósito de avaliar a degradação e o preenchimento físico (afinamento e espessamento) nas estruturas.
* **Abertura e Fechamento:** Foram tratadas falhas e imperfeições em textos digitalizados e impressões digitais ruidosas. A *Abertura* (através de `cv2.morphologyEx` com `cv2.MORPH_OPEN`) foi utilizada para remover impurezas locais (ruídos brancos no fundo preto), enquanto o *Fechamento* (`cv2.MORPH_CLOSE`) conectou trechos partidos das letras.
* **Gradiente Morfológico (Extração de Fronteiras):** O laboratório culminou na criação de um script de processamento contínuo da captura da Webcam. Utilizando `cv2.morphologyEx()` com `cv2.MORPH_GRADIENT`, extraiu-se o contorno dos objetos em tempo real. Para visualização comparativa, a imagem original e a extraída foram acopladas lateralmente usando `cv2.hconcat()` e o tamanho do kernel pôde ser alterado via teclado.

## 3. Análise e Discussão dos Estudos Realizados

**a) Erosão e Estrutura:** Ao aplicar uma erosão em uma imagem binária que contém ruídos brancos (pontos isolados de 1 pixel) utilizando um kernel 3x3 de *uns*, **os ruídos desaparecem completamente da imagem**. Isso ocorre porque, no funcionamento do operador de erosão, o pixel central só é mantido se **todos** os pixels sob o elemento estruturante forem iguais a 1 (forem brancos). Como o ruído tem apenas 1 pixel de tamanho e o kernel tem 3x3 (9 pixels), o kernel nunca caberá inteiramente dentro do ruído, fazendo com que o pixel seja convertido para o valor do fundo (0/preto).

**b) Dilatação e Conectividade:** Ao aplicar uma dilatação com um elemento estruturante pequeno (ex: 3x3) em um texto com letras levemente "quebradas" ou fragmentadas, o efeito é o **espessamento dos traços das letras**, o que acaba conectando as partes fragmentadas e restaurando a legibilidade dos caracteres. No entanto, se o kernel utilizado for grande demais, ocorre um **efeito colateral indesejado**: a expansão excessiva fará com que as letras vizinhas e as linhas de texto se fundam umas nas outras, transformando o texto em blocos sólidos e borrados, impossibilitando a leitura.

**c) Extração de Contornos (Gradiente):** A extração de fronteira é feita subtraindo a imagem erosionada da imagem original. Se aumentássemos o tamanho do elemento estruturante de 3x3 para 11x11, a erosão corroeria uma camada muito mais espessa das bordas do objeto original. Como resultado, ao subtrair essa imagem muito mais "encolhida" da original, **o contorno resultante ficaria consideravelmente mais grosso/largo**.

**d) Abertura vs. Fechamento:** * Para remover pequenos objetos indesejados (como poeira em um scan) sem alterar significativamente o tamanho dos objetos principais, utiliza-se a **Abertura** (Erosão seguida de Dilatação). A erosão elimina a poeira, e a dilatação restaura o tamanho original dos objetos maiores que sobreviveram à erosão.
* Para fechar pequenos buracos ou fendas dentro de um objeto sólido, utiliza-se o **Fechamento** (Dilatação seguida de Erosão). A dilatação expande as bordas e preenche os buracos, enquanto a erosão subsequente retorna as bordas externas do objeto ao seu tamanho original, mantendo o buraco preenchido.

**e) Análise de Kernel (Geometria):** Se aplicarmos uma dilatação em um único pixel branco no centro de uma imagem preta:
* Usando um **"kernel_cross"** (cruz), o formato do objeto resultante assumirá a exata forma de uma **cruz** com as dimensões do kernel.
* Usando um **"kernel_rect"** (retangular), o formato resultante será um **quadrado/retângulo** sólido (um bloco preenchido), correspondente às dimensões da matriz de uns.

## 4. Conclusões
O Laboratório 7 evidenciou que a morfologia matemática é indispensável para o tratamento estrutural de imagens binarizadas. Demonstrou-se que a escolha criteriosa do formato e tamanho do kernel dita o sucesso de operações vitais, como a eliminação de ruídos (Abertura) e a união de componentes fragmentados (Fechamento). 

Adicionalmente, verificou-se que a binarização prévia atua como o principal determinante para o sucesso das operações morfológicas, pois estas dependem de uma clara distinção lógica entre o "objeto" (foreground) e o "fundo" (background). A aplicação destes conceitos na captura em tempo real via Webcam confirmou que os operadores morfológicos, especialmente o gradiente, são altamente eficientes e adequados para aplicações práticas de visão computacional, ferramentas estas que poderão ser aplicadas na extração de características cruciais e correção de imperfeições no projeto final da equipe.
