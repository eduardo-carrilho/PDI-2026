## 1. Introdução

Este relatório apresenta as atividades e os resultados obtidos no Laboratório 5, cujo foco é o processamento de cores em imagens e vídeos. 

Os objetivos principais deste laboratório incluem a transição entre espaços de cores independentes e dependentes de dispositivos, a implementação de correções de ponto branco e a aplicação de transformações gama. Além disso, buscou-se manipular de forma prática modelos voltados para a percepção humana (como o CIELAB), para dispositivos (HSV) e para a compressão de vídeo (YCbCr). 

As implementações foram realizadas utilizando Python 3.x, NumPy, OpenCV e Matplotlib.

## 2. Procedimentos Experimentais

Nesta seção são detalhadas as operações realizadas ao longo da prática experimental, conforme os códigos desenvolvidos e testados no roteiro da disciplina.

* **2.1. Correção de Ponto Branco:** A técnica foi aplicada para ajustar a iluminação de uma imagem, garantindo que objetos brancos parecessem brancos sob diferentes fontes de luz. Utilizou-se a transformação de von Kries, dada por: $C_{new}=C_{old}(W_{target}/W_{source})$.
* **2.2. Transformação XYZ para RGB e Correção Gama:** Observou-se que o espaço XYZ é independente de dispositivo, mas para exibição em monitores, precisamos converter a imagem para RGB e aplicar a lei de potência (Gama).
* **2.3. O Modelo CIELAB ($L^*a^*b^*$):** A imagem foi convertida para o espaço CIELAB, que é projetado para ser perceptualmente uniforme. Os canais foram separados para observar que o canal $L^*$ representa a luminosidade, enquanto $a^*$ e $b^*$ são as dimensões de oponência de cor.
* **2.4. Cores Dependentes de Câmera (HSV e sRGB):** O espaço HSV foi utilizado por ser uma transformação não linear do RGB que facilita a segmentação de objetos por cor, sendo independente da intensidade luminosa.
* **2.5. Modelo YCbCr:** Foi realizada a conversão para o modelo YCbCr, que é essencial para compressão de vídeo (como MPEG e JPEG). Ele separa a Luminância (Y) da Crominância (Cb e Cr).
* **2.6. Aplicações em Tempo Real (Webcam):** Foram elaborados dois programas com captura de vídeo contínua:
    * O primeiro realiza continuamente sobre a imagem da WEBCAM as operações de correção de ponto branco e a correção gama, mostrando na tela o resultado em tempo real.
    * O segundo realiza a detecção de uma cor pelo método HSV na imagem da WEBCAM, mostrando lado a lado a imagem original e a imagem segmentada.
* **2.7. Resolução dos Exercícios Propostos:**
    * *Exercício 1:* Comparando a luminosidade, concluiu-se que o canal $L^*$ (CIELAB) preserva melhor os detalhes de sombreamento e contraste do que o canal Y (YCbCr). O $L^*$ é perceptualmente uniforme, enquanto o Y é uma transformação linear direta do RGB.
    * *Exercício 2:* A aplicação de um gama inverso ($\gamma=2.2$) escureceu consideravelmente a imagem. Os tons médios e escuros foram empurrados para perto de zero, causando a perda de detalhes (esmagamento das sombras).
    * *Exercício 3:* Ao zerar os canais Cb e Cr no espaço YCbCr e reconstruir a imagem para RGB, o filtro operou corretamente, resultando em uma imagem em escala de cinza.
    * *Exercício 4 (Desafio do Semáforo):* A segmentação utilizando HSV mostrou-se mais fácil de configurar porque a cor (matiz) está isolada no canal H. O espaço HSV lidou muito melhor com as áreas de sombra, pois a cor permanece relativamente constante mesmo quando a intensidade luminosa diminui.

## 3. Análise e Discussão dos Estudos Realizados

Abaixo são apresentadas as respostas analíticas sobre o processamento de cores:

**a) Ponto Branco e Constância de Cor:** A correção de ponto branco é considerada uma "transformação de adaptação cromática" pois simula o mecanismo do sistema visual humano. Ela reajusta os valores da imagem para que a percepção das cores se mantenha constante, independentemente das propriedades da fonte de luz.

**b) Espaços Independentes vs. Dependentes:** O espaço CIE XYZ é independente e descreve as cores com base na percepção humana padrão, enquanto o sRGB depende das limitações e características de reprodução do dispositivo (como os LEDs do monitor). Não exibimos imagens diretamente em XYZ porque os monitores precisam dessa conversão para mapear as cores corretamente em seu hardware.

**c) A Necessidade da Correção Gama:** Se a correção gama fosse omitida, as sombras da imagem sofreriam esmagamento visual e ela pareceria artificialmente escura. O gama compensa a resposta não linear dos monitores, evitando a perda de percepção de detalhes nas áreas escuras.

**d) Percepção Humana e o Modelo CIELAB:** Dizer que o modelo $L^*a^*b^*$ é "perceptualmente uniforme" significa que uma distância calculada matematicamente entre dois pontos de cor neste espaço corresponde de forma exata e proporcional à diferença visual percebida pelo olho humano.

**e) Engenharia de Vídeo e YCbCr:** O sistema visual humano tem alta resolução para detectar variações de luminosidade (Y), mas é ruim em perceber finas transições de cor (Cb e Cr). A compressão de vídeo (como JPEG ou MPEG) utiliza essa característica para reduzir a resolução (suavizar/borrar) os canais Cb e Cr, diminuindo drasticamente o tamanho dos arquivos sem causar perda de qualidade aparente.

**f) Representação Cilíndrica e Robustez:** O canal H (Hue) do HSV é frequentemente preferido porque isola a cor pura. Quando uma região colorida possui variações de iluminação, como sombras e reflexos, o RGB altera seus três canais simultaneamente, dificultando a segmentação. No HSV, a sombra altera majoritariamente apenas a intensidade no canal V (Value), deixando o canal H estável, o que garante maior robustez.

## 4. Conclusões

Este laboratório evidenciou a complexidade e a versatilidade necessárias para manipular imagens coloridas e vídeos. Conclui-se que a conversão entre espaços de cor é uma etapa obrigatória dependendo do objetivo do sistema: seja adaptar uma imagem para exibição correta via sRGB e Gama, preparar o processamento com CIELAB para medições fidedignas, ou comprimir dados de forma eficiente isolando a luminância da crominância no YCbCr. 

Destaca-se, por fim, a aplicação em segmentação computacional: a experimentação prática comprovou que espaços dependentes de iluminação, como o HSV, oferecem a robustez indispensável para sistemas de visão computacional lidarem com áreas de sombra e reflexos do mundo real.
