## 2. Fundamentação e Procedimentos Experimentais

A manipulação de cores exige a transição entre sistemas de coordenadas colorimétricas. Nesta etapa, implementamos as seguintes funções e transformações:

### 2.1 Correção de Ponto Branco (White Point Correction)
A correção de ponto branco visa ajustar a iluminação da cena para que objetos neutros sejam interpretados corretamente como branco, independentemente da temperatura de cor da fonte de luz. 
* **Procedimento:** Utilizou-se a transformação de *von Kries*, que aplica um escalonamento linear nos canais RGB baseando-se na razão entre o iluminante de destino (Target White) e o iluminante da fonte (Source White). 
* **Equação aplicada:** $C_{new} = C_{old} \times (W_{target} / W_{source})$.

### 2.2 Transformação XYZ e Correção Gama
O espaço CIEXYZ é um modelo independente de dispositivo que serve como base para quase todas as outras conversões. 
* **Procedimento:** A imagem foi convertida de RGB para XYZ. Em seguida, aplicou-se a **Correção Gama ($\gamma$)**, essencial para ajustar a resposta de luminância. Observou-se que valores de $\gamma < 1.0$ clareiam a imagem (expandindo os tons escuros), enquanto $\gamma > 1.0$ escurecem a composição, sendo um processo vital para a exibição fiel em diferentes monitores.

### 2.3 O Modelo CIELAB (L*a*b*)
O modelo L*a*b* é projetado para ser perceptualmente uniforme. 
* **Canal L (Luminosidade):** Representa o brilho de 0 a 100.
* **Canais a e b:** Representam as dimensões de oponência cromática (Verde-Vermelho e Azul-Amarelo).
* **Análise:** Este espaço permite alterar a iluminação da cena (canal L) sem causar distorções ou "estouros" nas cores originais, algo impossível de realizar diretamente no espaço RGB.

## 3. Segmentação e Rastreio de Objetos

Uma das aplicações mais críticas do processamento de cores é a segmentação. Comparamos a eficácia entre o modelo RGB e o modelo HSV.

### 3.1 Espaço de Cor HSV (Hue, Saturation, Value)
Diferente do RGB, o modelo HSV isola a informação da cor pura (Matiz/Hue) da sua intensidade (Saturação) e do seu brilho (Valor).
* **Vantagem Técnica:** Em tarefas de *tracking*, o canal **Hue** permanece estável mesmo sob variações de iluminação ou sombras. No experimento, definimos máscaras (ranges) para isolar cores específicas, demonstrando que a segmentação em HSV é significativamente mais robusta e intuitiva do que a tentativa de combinar limites nos três canais R, G e B simultaneamente.

### 3.2 Conversão para YCbCr
Este modelo é amplamente utilizado em padrões de compressão (JPEG e vídeo). Ele separa a Luminância (Y) das diferenças de cor azul (Cb) e vermelho (Cr). A análise mostrou como o sistema visual humano prioriza a nitidez do brilho em detrimento da cor, permitindo técnicas de subamostragem cromática para redução de tamanho de arquivos.

## 4. Análise e Discussão dos Resultados

Durante a execução dos códigos no arquivo `pdi26__lab5__funcoes_cores_.ipynb`, os seguintes pontos foram destacados:

* **Balanço Cromático:** A correção de ponto branco provou ser o primeiro passo essencial em qualquer pipeline de visão computacional; sem ela, algoritmos de segmentação baseados em cor fixa falham ao mudar a lâmpada do ambiente.
* **Robustez em HSV:** A comparação visual das máscaras de segmentação revelou que o método HSV lida muito melhor com sombras e reflexos. Enquanto a máscara RGB apresentava furos ou ruídos em áreas de transição, a máscara HSV manteve a integridade da forma do objeto segmentado.
* **Percepção e Brilho:** A manipulação do espaço CIELAB confirmou que a percepção humana de brilho é mais precisa quando separada da informação cromática, validando o uso desse modelo em tratamentos de imagem profissionais.

## 6. Conclusão

Neste laboratório, consolidou-se o entendimento de que a cor é uma informação multidimensional que deve ser tratada de acordo com o objetivo do software. 

As principais conclusões foram:
1. **Escolha do Modelo:** O espaço **RGB** deve ser restrito à aquisição e exibição. Para processamento analítico, deve-se transitar para **HSV** ou **CIELAB**.
2. **Importância da Pré-Calibração:** A correção gama e o ponto branco não são apenas ajustes estéticos, mas correções físicas necessárias para que os dados numéricos da imagem correspondam à realidade física.
3. **Eficiência de Compressão:** O estudo do YCbCr evidenciou como a engenharia de multimídia aproveita as limitações biológicas do olho humano para otimizar o armazenamento de dados.

O experimento forneceu a base técnica necessária para futuras implementações de reconhecimento facial, análise médica e compressão de vídeo, onde o controle da cor é o fator determinante da precisão do sistema.
